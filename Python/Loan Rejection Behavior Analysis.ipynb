## Handle Missing Values

```python
# Income → median
df['Income'].fillna(df['Income'].median(), inplace=True)

# Credit Score → median
df['Credit_Score'].fillna(df['Credit_Score'].median(), inplace=True)
```

## Feature Engineering

In this step, we transform cleaned data into business-relevant features that help analyze loan approval and rejection behavior.

This step is responsible for:

- Creating financial risk indicators
- Segmenting customers
- Building analytical features for SQL & dashboards

This forms the business logic layer of the analytics pipeline.

### Income to Loan Ratio

```python
df['Income_to_Loan_Ratio'] = df['Income'] / df['Loan_Amount']

df['Income_to_Loan_Ratio'] = df['Income_to_Loan_Ratio'].round(3)
```

### Credit Score Segmentation

```python
df['Credit_Category'] = pd.cut(
    df['Credit_Score'],
    bins=[300, 600, 700, 950],
    labels=['Poor', 'Average', 'Good']
)
```

### Debt Burden Category

```python
max_val = df['Debt_to_Income_Ratio'].max()

# Ensure last bin is always greater than 1
upper = max(max_val, 1) + 0.01

df['debt_burden'] = pd.cut(
    df['Debt_to_Income_Ratio'],
    bins=[0, 0.3, 0.6, 1, upper],
    labels=['Low', 'Medium', 'High', 'Very High'],
    include_lowest=True
)
```

```python
df.head()
```

### Age Group Segmentation

```python
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[20,30,40,50,60],
    labels=['20-30','30-40','40-50','50-60']
)
```

```python
df.head()
```

### Risk Score

```python
df['Risk_Score'] = (
    (850 - df['Credit_Score']) / 550 * 50 +   # normalized credit risk
    df['Debt_to_Income_Ratio'] * 50                            # DTI contribution
)
df['Risk_Score'] = df['Risk_Score'].round(1)
```

```python
df.head()
```

### High Risk Flag

```python
df['High_Risk'] = np.where(df['Risk_Score'] > 50, 1, 0)
```
